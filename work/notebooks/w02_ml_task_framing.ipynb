## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Answer :
A fixed rule is rigid and fails to account for the nuance of different content types. For example, a simple rule might flag all long-form articles as 'high potential,' even if they are purely informational and don't require the same engagement as a transactional page. An ML model beats a fixed rule because it learns non-linear, multi-dimensional interactions. It doesn't just look at character count in isolation; it learns how the impact of character count changes depending on the main_intent, avg_position, and content_type. This allows the model to create a 'click expectation' specific to the context of each individual page, which is impossible to capture with a static, hand-written rule
